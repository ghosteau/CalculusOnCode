# Central Limit Theorem Simulation

### The theorem itself states the following:
The Central Limit Theorem (CLT) states that if you take many random samples from a population, the distribution of their sample means approaches a normal distribution—regardless of the population’s original shape. 

If you want to learn more, you can check out *this* Investopedia article: https://www.investopedia.com/terms/c/central_limit_theorem.asp

**What's our approach?**

For this example, we will do the most trivial example: dice rolls. We can do this all simply with the random library in Python, NumPy (to speed things up ever-so-slighty) and matplotlib to verify our findings (e.g. is it really going to look like a normal distribution or not?)

*Note: on your own, feel free to uncomment any of the print statements I commented out. This was just to keep the notebook non-verbose, but uncommenting can be a great way to see how the functions work in the case that you are confused by the logic.*

```python
import random

import numpy as np
import matplotlib.pyplot as plt

# Suppose dice rolls, discrete uniform distribution
def simulate_dice(rolls):
    results = []
    for _ in range(rolls):
        roll = random.randint(1, 6)
        results.append(roll)
    
    return results

# This simulates our sample means
# Here, n is the number of dice rolled in each simulation, and k is the number of simulations we run to build the distribution of sample means.
def simulate_means(n, k):
    means = []
    for _ in range(k):
        sim = simulate_dice(n)
        mean = np.mean(sim)
        means.append(mean)
        
    return means
```

```python
m = 2000
# print(f"Result of {m} dice rolls:")
dice = simulate_dice(m)
# print(dice)
# print()

n = 10000
k = 2500
means = simulate_means(n, k)
# print(f"Mean of {n} dice rolls, {k} simulations:")
# print(means)
```

```python
means = [float(x) for x in means]

# Note that, by the definition of statistical expectation, we would expect the average of a 6 sided dice roll to be 3.5
plt.figure(figsize=(8, 5))
plt.hist(means, bins=20, edgecolor='black', alpha=0.7)
plt.axvline(np.mean(means), color='red', linestyle='--', linewidth=2, label=f'Mean = {np.mean(means):.2f}')

plt.title(f"Distribution of sample means")
plt.xlabel(f"Sample mean")
plt.ylabel("Frequency")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
